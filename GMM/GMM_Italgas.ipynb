```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_path = '../data/faulty_data.parquet'
data = pd.read_parquet(data_path)
```

```python

# Scatter plot colored by CaseId (categorical string)
plt.figure(figsize=(8, 6))
# Convert CaseId to categorical codes for scatter coloring
caseid_codes, uniques = pd.factorize(data['CaseId'])
scatter = plt.scatter(
    data['AverageFlowRate'],
    data['PeakCount'],
    c=caseid_codes,
    cmap='tab10',
    alpha=0.2
)
plt.xlabel('AverageFlowRate')
plt.ylabel('PeakCount')
plt.title('AverageFlowRate vs PeakCount (colored by CaseId)')

# Create a legend for the categorical color mapping
handles = [
    plt.Line2D([0], [0], marker='o', color='w', alpha=0.4,
               label=label, markerfacecolor=plt.cm.tab10(i % 10), markersize=8)
    for i, label in enumerate(uniques)
]
plt.legend(handles=handles, title='CaseId', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
